# 🎙️ مصنع الصوت - بضغطة واحدة بدون كتابة

```
%cd /content
!rm -rf /content/my-voice-clone
!mkdir -p /content/my-voice-clone
%cd /content/my-voice-clone

with open('requirements.txt','w') as f:
    f.write('TTS==0.22.0\ngradio==4.44.0\ntorch\ntorchaudio\n')

code = '''
import gradio as gr
from TTS.api import TTS
print('⏳ تحميل XTTS-v2...')
tts = TTS('tts_models/multilingual/multi-dataset/xtts_v2')
print('✅ جاهز')
def clone(text, wav, lang):
    if wav is None: return None, 'ارفع عينة'
    if not text.strip(): return None, 'اكتب نص'
    out='/tmp/out.wav'
    tts.tts_to_file(text=text, speaker_wav=wav, language=lang, file_path=out)
    return out, 'تم ✅'
with gr.Blocks(title='مصنع الصوت') as demo:
    gr.Markdown('# 🎙️ مصنع الصوت XTTS-v2 - يعمل بضغطة واحدة')
    wav=gr.Audio(type='filepath', label='عينة صوتك WAV 6 ثواني')
    lang=gr.Dropdown(['ar','en'], value='ar', label='اللغة')
    txt=gr.Textbox('السلام عليكم هذا صوتي المستنسخ', label='النص')
    btn=gr.Button('إنتاج', variant='primary')
    out=gr.Audio(label='النتيجة')
    st=gr.Textbox(label='الحالة')
    btn.click(clone, [txt, wav, lang], [out, st])
demo.launch(share=True, server_name='0.0.0.0', server_port=7860)
'''
with open('app.py','w', encoding='utf-8') as f:
    f.write(code)

!pip install -q -r requirements.txt
!python app.py
```
